```python
import os
import boto3
from botocore.client import Config

# Configurações do MinIO/S3
endpoint_url = "http://minio:9000"  # ou "http://minio:9000" se rodando dentro do mesmo docker network
access_key = "minioadmin"
secret_key = "minioadmin"
bucket_name = "datalake"

# Inicializa o cliente S3
s3 = boto3.client(
    "s3",
    endpoint_url=endpoint_url,
    aws_access_key_id=access_key,
    aws_secret_access_key=secret_key,
    config=Config(signature_version="s3v4"),
    region_name="us-east-1"
)

# Caminho local da pasta data
base_dir = os.path.abspath("../data")

lake_layer='raw_data'

for root, dirs, files in os.walk(base_dir):
    for file in files:
        local_path = os.path.join(root, file)
        # Gera o caminho relativo para manter a estrutura de pastas
        relative_path = os.path.relpath(local_path, base_dir)
        s3_key = f'{lake_layer}/{relative_path}'
        if '.csv' in s3_key:
            print(s3_key)
            s3.upload_file(local_path, bucket_name, s3_key)
```
